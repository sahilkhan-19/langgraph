```python
from dotenv import load_dotenv
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Literal

load_dotenv()
```

```python
class QuadState(TypedDict):
    a: int
    b: int 
    c: int

    equation: str
    discriminant: float
    result: str
```

```python
def show_eqn(state: QuadState):

    eqn = f'{state["a"]}x^2 {state["b"]}x {state["c"]}'
    return {'equation': eqn}

def cal_dis(state: QuadState):

    dis = (state["b"]**2 - 4*state["a"]*state["c"])

    return {'discriminant': dis}

def real_roots(state: QuadState):

    root1 = (-state["b"] + state["discriminant"] ** 0.5)/(2*state["a"])
    root2 = (-state["b"] - state["discriminant"] ** 0.5)/(2*state["a"])

    result = f'The roots are {root1} and {root2}'

    return {'result': result}

def repeated_roots(state: QuadState):

    root = (-state["b"])/(2*state["a"])

    result = f'The only repeating roots is {root}'

    return {'result': result}

def no_real_roots(state: QuadState):

    result = f'No real roots'

    return {'result': result}

def check_condn(state: QuadState) -> Literal["real_roots", "repeated_roots", "no_real_roots"]:
    if state['discriminant'] > 0:
        return "real_roots"
    elif state['discriminant'] == 0:
        return "repeated_roots"
    else: 
        return "no_real_roots"
```

```python
graph = StateGraph(QuadState)

graph.add_node('show_eqn', show_eqn)
graph.add_node('cal_dis', cal_dis)
graph.add_node('real_roots', real_roots)
graph.add_node('repeated_roots', repeated_roots)
graph.add_node('no_real_roots', no_real_roots)

graph.add_edge(START, 'show_eqn')
graph.add_edge('show_eqn', 'cal_dis')
graph.add_edge('cal_dis', END)
graph.add_conditional_edges('cal_dis', check_condn)
graph.add_edge('real_roots', END)
graph.add_edge('repeated_roots', END)
graph.add_edge('no_real_roots', END)

workflow = graph.compile()
workflow
```

```python
initial_state = {
    'a': 4,
    'b': -5,
    'c': -4
}

workflow.invoke(initial_state)
```
